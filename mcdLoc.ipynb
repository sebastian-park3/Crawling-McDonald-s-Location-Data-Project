```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time


# User inputs city
city = input("Enter city and state (e.g., New York, NY): ")


# Open URL on Chrome
mcdUrl = 'https://www.mcdonalds.com/us/en-us/restaurant-locator.html'
driver = webdriver.Chrome()
driver.get(mcdUrl);


# WebDriverWait from selenium
wait = WebDriverWait(driver, 10)


# Wait for search box and input user-entered city into said search box
inputFormula = wait.until(EC.presence_of_element_located((By.ID, "form-text-1673594539")))
inputFormula.send_keys(city)
inputFormula.send_keys(Keys.ENTER)


# Wait until at least one result appears
wait.until(EC.presence_of_element_located((By.CLASS_NAME, "cmp-restaurant-locator__restaurant-list-item")))


# Initialize moreBtn and raw variables. raw is the count of remaining restaurants shown in the "Load More" button
moreBtn = driver.find_element(By.ID, "button-93a5672f17")
count_span = moreBtn.find_element(By.CSS_SELECTOR, "span.cmp-restaurant-locator__restaurant-load-more-count")
raw = count_span.text.strip()


# Click "Load More" button until all restaruants are visible
while not raw == '(-3)':
    try:
        moreBtn = driver.find_element(By.ID, "button-93a5672f17")
        count_span = moreBtn.find_element(By.CSS_SELECTOR, "span.cmp-restaurant-locator__restaurant-load-more-count")
        raw = count_span.text.strip() 
        moreBtn.click()
        time.sleep(0.5)
    except (TimeoutException, Exception):
        break


# Get locations of restaurant result WebDriver objects and put them into a list
storeAddresses = driver.find_elements(By.CSS_SELECTOR, ".cmp-restaurant-locator__restaurant-list-item-details-al2")


# Get streets of restaurant result WebDriver objects and put them into a list
streets = driver.find_elements(By.CSS_SELECTOR,".cmp-restaurant-locator__restaurant-list-item-details-al1")


# Make lists of text within WebDriver objects and put them into lists
street_texts = [el.text.strip() for el in streets]
address_texts = [el.text.strip() for el in storeAddresses]


# Close Chrome
driver.close()


# Print locations
print(f'Here are the McDonald\'s locations closest to {city}:')
for street, addr in zip(street_texts, address_texts):
    print(f"{street}, {addr}")
```
